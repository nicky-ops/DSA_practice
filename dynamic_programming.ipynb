### Dynamic Programming
Most powerful design technique for solving optimization problems and is based on the intuition of the divide-and-conquer technique. In this case we explore the space of all the possible solutions by decomposing the problem into a series of sub-problems and then combining the results to compute the correct solution for the large problem.
This technique is used when the sub-problems are overlapping, meaning that the sub-problems share sub-sub-problems(each sub-sub-problems is solved only once and solutions to an already encountered sub-problem is not recomputed) unlike in divide-and-conquer which solves a problem by combining the solutions of non-overlapping(disjoint) sub-problems(each sub-problem has to be solved before its results can be used to solve bigger problems).

#### Characteristics
* **Optimal substructure**: Given any problem, if the solution can be obtained by combining the solutions of its sub-problems, then the problem is said to have an optimal substructure.
* **Overlapping sub-problem**: if an algorithm has to repeatedly solve the same sub-problem again and again, then the problem has overlapping sub-problems.

If a problem has these x-tics then dynamic programming is useful, since the implementation can be improved by reusing the same solution computed before.
In this approach we need an efficient way to store the results of each sub-problem. The following techniques are used:
1. **Top-down with memoization**: Memoization refers to storing the solution of the sub-problem in an array or hash table. Whenever a solution to a sub-problem needs to be computed, it is first referred to the saved values if it is already computed, and if it is not stored, then it is computed in the usual manner. This is called *memoized*, which means it "remembers" the results of the operation that has been computed before.

2. **Bottom-up approach**: This approach depends upon the "size" of the sub-problems. We solve smaller sub-problems first, and then while solving a particular sub-problem, we already have a solution of the smaller sub-problems on which it depends. Each sub-problem is solved only once, and whenever we try to solve any sub-problem, solutions to all the prerequisite smaller sub-problems are available, which can be used to solve it.

##### 1. Calculating the fibonacci series
Fibonacci series is a sequence in which each element is the sum of the two preceding elements.
```
func(0) = 1
func(1) = 1
func(n) = func(n-1) + func(n-2) for n > 1
```
Example to generate to the fifth term
`1 1 2 3 5`

````python
# without memoization
def fib(n):
    if n <= 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)

fib_list = []
for i in range(5):
    fib_list.append(fib(i))
print(fib_list)
````

### Greedy algorithms
Involve optimization and combinatorial problems, with the objective of obtaining the optimum solution from many possible solutions in each step. The aim is to get the local optimum solution, which may lead to obtaining the global optimum solution.
Standard problems where greedy algorithms can be used:
* Kruskal's minimum spanning tree.
* Dijkstra's shortest path problem.
* The Knapsack problem.
* Prim's minimal spanning tree algorithm.
* The travelling salesperson problem.

#### Shortest path problem
Requires one to find the shortest possible route between nodes on a graph. 
Dijkstra's algorithm is used to solve this problem. It works for weighted directed and undirected graphs. It produces a list of the shortest paht from a given source node, A, in a weighted graph.
