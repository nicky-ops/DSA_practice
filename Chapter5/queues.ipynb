# Queues
A list of elements stored in sequence where:
1. Data elements can only be inserted from one end, the rear end/tail of the queue.
2. Data elements can only be deleted from the other end, the front end of the queue
3. Only data elements from the front of the queue can be read.
- **enqueue** - operation to add an element to the queue
- **dequeue** - operation to remove an element fron the queue.
- Follows FIFO(first in first out) policy

## Python's list-based queues
Limitation of this implementation is the length of the queue is fixed.

```python
class ListQueue:
    def __init__(self):
        self.items = []
        self.front = self.rear = 0
        self.size = 3

    def enqueue(self, data):
        if self.size == self.rear:
            print("\n Queue is full")
        else:
            self.items.append(data)
            self.rear += 1

    def dequeue(self):
        if self.front == self.rear:
            print("Queue is empty")
        else:
            data = self.items.pop(0)
            self.rear -= 1
            return data
        
```

```python
q = ListQueue()
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)

print(q.items)
```

```python
data = q.dequeue()
print(data)
print(q.items)
```

## Linked list based queues

```python
class Node:
    def __init__(self, data=None, next=None, prev=None):
        self.data = data
        self.next = next
        self.prev = prev


class Queue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.count = 0

    def enqueue(self, data):
        new_node = Node(data, None, None)
        if self.head == None:
            self.head = new_node
            self.tail = self.head
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

        self.count += 1

    def dequeue(self):
        if self.count == 1:
            self.count -= 1
            self.head = None
            self.tail = None
        elif self.count > 1:
            self.head = self.head.next
            self.head.prev = None
        elif self.count < 1:
            print("Queue is empty")

        self.count -= 1
```

## Stack-based Queues
### Approach 1: When dequeue operation is costly
We use two stacks(Stack-1, Stack-2) for the implementation of the queue.
Steps:
1. Elements are removed from Stack-1, and then one by one all the elements are added to Stack-2.
2. The topmost element will be popped off Stack-2 and will be returned as the desired element.
3. Finally, the remaining elements are popped off Stack-2 one by one and then pushed again to Stack-1.

The worst-case time complexity of the enqueue operations is O(1), since any element can be added directly to the first stack.
The time complexity of the dequeue operation is O(n), since all elements are accessed and transferred from Stack-1 to Stack-2.

### Approach 2: When the enqueue operation is costly
Steps:
1. Move all the elements from Stack-1 to Stack-2.
2. Push the element we want to enqueue to Stack-2.
3. Move all the elements from Stack-2 to Stack-1 one by one. Pop the elements from Stack-2 and push them to Stack-1
The time complexity for the enqueue operation is O(n), and for the dequeue operation, it is O(1)

```python
# Approach 1: Costly Dequeue Operation
class Queue:
    def __init__(self):
        self.Stack1 = []
        self.Stack2 = []

    def enqueue(self, data):
        self.Stack1.append(data)

    def dequeue(self):
        if not self.Stack2:
            while self.Stack1:
                self.Stack2.append(self.Stack1.pop())
        if not self.Stack2:
            print("No element to dequeue")
            return
        return self.Stack2.pop()
```

```python
queue = Queue()
queue.enqueue(23)
queue.enqueue(13)
queue.enqueue(11)
print(queue.Stack1)
```

```python
queue = Queue()
queue.enqueue(23)
queue.enqueue(13)
queue.enqueue(11)
print(queue.Stack1)

queue.dequeue()
print(queue.Stack2)
```

# Applications of queues
1. Printing - many computers can share one printer. When the printer is ready to print, it will pick one of the jobs in the queue to print out. It will print the command from the computer that has given the command first and will choose the following jobs in order in which they were submitted by the different computers.
2. Operating systems queue processes to be executed by the CPU.
3. Music players implement queues to play songs in order

```python
from random import randint

class Track:
    def __init__(self, title=None):
        self.title = title
        self.length = randint(5, 10)
        self.count = 0
```

```python
track1 = Track("Whistle whistle")
track2 = Track("butter butter")
print(track1.length)
print(track2.length)
```

```python
import time
class MediaPlayerQueue(Queue):
    def add_track(self, track):
        self.enqueue(track)
        self.count += 1

    def play(self):
        while self.count > 0:
            current_track_node = self.dequeue()
            print("Now playing {}".format(current_track_node.data.title))
            time.sleep(current_track_node.data.length)
```

```python
track1 = Track("white whistle")
track2 = Track("butter butter")
track3 = Track("Oh black star")
track4 = Track("Watch that chicken")
track5 = Track("Don't go")
```

```python
print(track1.length)
print(track2.length)
```

```python
media_player = MediaPlayerQueue()
media_player.add_track(track1)
media_player.add_track(track2)
media_player.add_track(track3)
media_player.add_track(track4)
media_player.add_track(track5)
media_player.play()
```
