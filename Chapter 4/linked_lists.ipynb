# Chapter 4: Linked List
Python's built-in implementation of a list data structure is used to store data using a linked list.

**Linked list** - a data structure that stores elements in a linear order. It uses pointer structures to provide efficient storage of data in linear order.

Pointers are used to store the memory address of data items. They store the data and location, and the location stores the position of the next data item in the memory.

#### 1. Arrays
A collection of data items of the same type, stored in contiguous memory locations. This makes it easier and faster to compute the position of any element in the array using **offset** (an integer indicating the displacement between the first element and a given element) and **base address** (address of the memory location where the first element is stored).

|--|--|--|--|--|--|--|
| 3|11|7 |1 |4 |2 |1 |

##### Limitations
1. If the data to be stored in the array is large and the system has low memory, it is difficult to allot a large block of memory locations.
2. It has a static size that has to be declared at the time of creation.
3. Insertion and deletion operations in array data structures are slow as compared to linked lists - it is difficult to insert an element in an array at a given location since all data elements after the desired position must be shifted and then new elements inserted in between.

Array data structures are suitable when we want to do a lot of accessing of elements and fewer insertion and deletion operations, whereas linked list are suitable in applications where the size of the list is not fixed, and a lot of insertion and deletion operations will be required.
