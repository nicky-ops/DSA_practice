# Chapter 4: Linked List
Python's built-in implementation of a list data structure is used to store data using a linked list.

**Linked list** - a data structure that stores elements in a linear order. It uses pointer structures to provide efficient storage of data in linear order.

Pointers are used to store the memory address of data items. They store the data and location, and the location stores the position of the next data item in the memory.

### 1. Arrays
A collection of data items of the same type, stored in contiguous memory locations. This makes it easier and faster to compute the position of any element in the array using **offset** (an integer indicating the displacement between the first element and a given element) and **base address** (address of the memory location where the first element is stored).
| | | | | | | |
|---|---|---|---|---|---|---|
| 3|11|7 |1 |4 |2 |1 |

##### Limitations
1. If the data to be stored in the array is large and the system has low memory, it is difficult to allot a large block of memory locations.
2. It has a static size that has to be declared at the time of creation.
3. Insertion and deletion operations in array data structures are slow as compared to linked lists - it is difficult to insert an element in an array at a given location since all data elements after the desired position must be shifted and then new elements inserted in between.

Array data structures are suitable when we want to do a lot of accessing of elements and fewer insertion and deletion operations, whereas linked list are suitable in applications where the size of the list is not fixed, and a lot of insertion and deletion operations will be required.

### 2. Linked Lists
Properties:
1. The data elements are stored in memory in different locations that are connected through pointers. A pointer is an object that can store the memory address of a variable, and each data element points to the next data element until the last element which points to None
2. The length of the list can increase or decrease during execution of the program.

A node is a data container that contains the actual data and a pointer to one or more nodes.

```python
class Node:
    '''
    A very peculiar node
    '''
    def __init__(self, data=None):
        '''
        Initialize variables when creating an instance of the class
        '''
        self.data = data
        self.next = None


node1 = Node("eggs")
node2 = Node("ham")
node1.next = node2

print(node1.next, node2)
```

#### Singly linked lists
A singly linked list contains a number of nodes in which each node contains data and a pointer that links to the next node. The link of the last node in the list is None.

##### Creating and Traversing

```python
n1 = Node("Piastri")
n2 = Node("Hamilton")
n3 = Node("Butter")

n1.next = n2
n2.next = n3

# traversing the list
current = n1
while current:
    print(current.data)
    current = current.next
```

```python
# Using an iterator
def iter(self):
    current = self.head
    while current:
        val = current.data
        current = current.next
        yield val
```

```python
class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0

    def append(self, data):
        # Encapsulate the data in a Node
        node = Node(data)
        if self.head is None:
            self.head = node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = node
```

```python
words = SinglyLinkedList()
words.append('egg')
words.append('ham')
words.append('spam')
```

```python
current = words.head
while current:
    print(current.data)
    current = current.next
```

```python
# Efficient implementation where we track the last node of the list reducing worst-case running time of the append operation from 0(n) to 0(1)
class SinglyLinkedList:
    def __init__(self):
        self.tail = None
        self.head = None
        self.size = 0

    def append(self, data):
        node = Node(data)
        if self.tail:
            self.tail.next = node
            self.tail = node
        else:
            self.head = node
            self.tail = node
```

```python
words = SinglyLinkedList()
words.append('rich')
words.append('ham')
words.append('kaleb')

current = words.head
while current:
    print(current.data)
    current = current.next
```

##### Appending items at intermediate positions
To insert an element in at a given position, we traverse the list to reach the desired position where we want to insert an element. An element can be inserted in between two successive nodes using two pointers (prev and current)

```python
class SinglyLinkedList:
    def __init__(self):
        self.tail = None
        self.head = None
        self.size = 0

    def append(self, data):
        node = Node(data)
        if self.tail:
            self.tail.next = node
            self.tail = node
        else:
            self.head = node
            self.tail = node

    def append_at_a_location(self, data, index):
        current = self.head
        prev = self.head
        node = Node(data)
        count = 1
        while current:
            if count == 1:
                node.next = current
                self.head = node
                return
            elif count == index:
                node.next = current
                prev.next = node
                return
            count += 1
            prev = current
            current = current.next
        if count < index:
            print("This list has less number of elements")
```

```python
words = SinglyLinkedList()
words.append('egg')
words.append('ham')
words.append('spam')
current = words.head

while current:
    print(current.data)
    current = current.next

words.append_at_a_location('new', 2)
current = words.head
while current:
    print(current.data)
    current = current.next
```
